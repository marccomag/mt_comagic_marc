## Resample rasters to size of NTL

### Pop Count 100m

```python
from osgeo import gdal

gdal.UseExceptions()

# Paths
ntl_path = r"C:\Users\mcoma\OneDrive - Universität Zürich UZH\Masterarbeit\Data\NASA\output\nighttime_lights_nepal.tif"
pop_path = r"C:\Users\mcoma\OneDrive - Universität Zürich UZH\Masterarbeit\Data\population_count\npl_pop_2026_CN_100m_R2025A_v1_fixed.tif"
pop_resampled_path = r"C:\Users\mcoma\OneDrive - Universität Zürich UZH\Masterarbeit\Data\population_count\pop_resampled_to_ntl.tif"

# Open NTL reference
ref = gdal.Open(ntl_path)
ref_proj = ref.GetProjection()
ref_gt = ref.GetGeoTransform()
ref_cols = ref.RasterXSize
ref_rows = ref.RasterYSize

xmin = ref_gt[0]
ymax = ref_gt[3]
xmax = xmin + ref_cols * ref_gt[1]
ymin = ymax + ref_rows * ref_gt[5]

print(f"NTL Extent: {xmin}, {ymin}, {xmax}, {ymax}")
print(f"NTL Size: {ref_cols} x {ref_rows} Pixel")

# Resample population raster to match NTL raster
gdal.Warp(
    pop_resampled_path,
    pop_path,
    format='GTiff',
    outputBounds=(xmin, ymin, xmax, ymax),
    width=ref_cols,
    height=ref_rows,
    resampleAlg='sum',
    dstSRS=ref_proj
)

print("Done: pop_resampled_to_ntl.tif")
```
